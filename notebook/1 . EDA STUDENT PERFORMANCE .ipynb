```python
Q1 = df["Units Sold"].quantile(0.25)
Q3 = df["Units Sold"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = df[(df["Units Sold"] < lower) | (df["Units Sold"] > upper)]
print("Outliers:", len(outliers))
print("Outlier %:", len(outliers)/len(df)*100)
print("Upper bound:", upper)
```

```python
from scipy.stats import f_oneway

a = df[df["Seasonality"]=="Autumn"]["Units Sold"]
s = df[df["Seasonality"]=="Summer"]["Units Sold"]
w = df[df["Seasonality"]=="Winter"]["Units Sold"]
sp = df[df["Seasonality"]=="Spring"]["Units Sold"]

f_stat, p = f_oneway(a, s, w, sp)
print("p-value:", p)
```

```python
df['Month'] = df['Date'].dt.to_period('M')
monthly_df = df.groupby('Month')[['Units Sold', 'Demand Forecast']].sum().reset_index()
monthly_df['Month'] = monthly_df['Month'].dt.to_timestamp()
```

```python
plt.figure(figsize=(12,6))
sns.lineplot(x='Month', y='Units Sold', data=monthly_df, label='Actual Sales', marker='o')
sns.lineplot(x='Month', y='Demand Forecast', data=monthly_df, label='Forecasted Demand', marker='s')

plt.title('Actual Sales vs Forecasted Demand (Month-wise)', fontsize=14)
plt.xlabel('Month')
plt.ylabel('Units')
plt.legend()
plt.grid(True)
plt.show()
```
